```python
import matplotlib.pyplot as plt
import os
import numpy as np
import pandas as pd
```

```python
labeled_df = pd.read_csv('../processed/final_labeled_data.csv')
cleaned_df = pd.read_csv('../processed/input_cleaned_w1_2_3.csv')
```

```python
def draw_field(ax):
    ax.set_xlim(0, 120)
    ax.set_ylim(0, 53.3)
    ax.set_aspect('equal')
    ax.set_facecolor("green")
    ax.set_xticks([])
    ax.set_yticks([])

    # Yard lines
    for x in range(0, 121, 10):
        ax.axvline(x, color="white", alpha=0.3)
```

```python
def draw_frame(frame_df, play_id, frame_id, save_path):
    fig, ax = plt.subplots(figsize=(12, 6))
    draw_field(ax)

    offense = frame_df[frame_df["player_side"] == "Offense"]
    defense = frame_df[frame_df["player_side"] == "Defense"]

    # Plot players
    ax.scatter(offense["x"], offense["y"], c="blue", s=40, label="Offense")
    ax.scatter(defense["x"], defense["y"], c="red", s=40, label="Defense")

    # Ball (average of landing or ball carrier if present)
    if "ball_land_x" in frame_df.columns:
        bx = frame_df["ball_land_x"].iloc[0]
        by = frame_df["ball_land_y"].iloc[0]
        ax.scatter(bx, by, c="brown", s=80, marker="o")

    ax.set_title(f"Play {play_id} | Frame {frame_id}")
    plt.legend(loc="upper right")

    os.makedirs(save_path, exist_ok=True)
    plt.savefig(f"{save_path}/frame_{frame_id:04d}.png", dpi=150)
    plt.close()
```

```python
def generate_frames_for_play(df, play_id, output_dir="frames"):
    play_df = df[df["play_id"] == play_id]

    for frame_id, frame_df in play_df.groupby("frame_id"):
        save_path = f"{output_dir}/play_{play_id}"
        draw_frame(frame_df, play_id, frame_id, save_path)
```

```python
labeled_plays = labeled_df["play_id"].unique()

for pid in labeled_plays:
    generate_frames_for_play(cleaned_df, pid)
```
