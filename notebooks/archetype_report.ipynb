```python
import pandas as pd
import numpy as np
from scipy.stats import ttest_ind

#Load Dataset
df = pd.read_csv("../processed/modeling_dataset.csv")


#Create Data-Driven Quick Instinct Archetype 
#QI = players in the top 25% of Eyes AND Innovation

eyes_q = df["Eyes"].quantile(0.75)
innov_q = df["Innovation"].quantile(0.75)

df["archetype"] = np.where(
    (df["Eyes"] >= eyes_q) & (df["Innovation"] >= innov_q),
    "QI",
    "Other"
)

print("Archetype counts:")
print(df["archetype"].value_counts())


#Split groups for PER10 comparison

qi = df[df["archetype"] == "QI"]["PER10"]
non_qi = df[df["archetype"] != "QI"]["PER10"]


#Perform statistical testing
# One-tailed independent samples t-test

t_stat, p_two = ttest_ind(qi, non_qi, equal_var=False)
p_one = p_two / 2  # convert two-tailed to one-tailed

#Effect size: Cohen's d
d = (qi.mean() - non_qi.mean()) / np.sqrt(
    ((qi.std()**2 + non_qi.std()**2) / 2)
)


# Display results
print('\n')
print("QI count:", len(qi))
print("Other count:", len(non_qi))
print("QI mean PER10:", round(qi.mean(), 2))
print("Other mean PER10:", round(non_qi.mean(), 2))
print("t-stat:", round(t_stat, 4))
print("p-value (one-tailed):", round(p_one, 6))
print("Cohen's d:", round(d, 3))
```
